```python
# 1. Cargamos tu dimensión desde el archivo local de la carpeta
 
# Reconstruimos el diccionario asegurando que eliminamos espacios en blanco traicioneros

map_territorios = {str(row['nombre_territorio']).strip(): int(row['id_territorio']) for _, row in df_territorio.iterrows()}

print(map_territorios)

 
registro_empresas = {}
 
# 2. Procesamos el JSON que ya tienes en memoria en la variable 'data'

for serie in data:
    
    nombre_completo = str(serie['Nombre'])
    
    nombre_lower = nombre_completo.lower()
    print(nombre_lower)

    # Identificar el territorio mediante búsqueda flexible por contenido de texto

    id_terr = None

    for nombre_terr, id_t in map_territorios.items():

        if nombre_terr.lower() in nombre_lower:

            id_terr = id_t
            print("id_terr", id_terr)
            break

    # Forzar la captura del total del país si la frase contiene la palabra nacional

    if "nacional" in nombre_lower:

        id_terr = map_territorios.get('Nacional')
        print("nacional", id_terr)

    # Si la serie pertenece a una provincia o municipio que no está en territorio.csv, pasamos

    if id_terr is None:

        continue

    # 3. Procesamos los históricos mensuales correspondientes

    for dato in serie['Data']:

        valor = dato.get('Valor')

        print("valor", valor)

        id_time = dato.get('CodigoPeriodo')

        print("id_time", id_time)

        # Filtro: nos quedamos únicamente con los códigos mensuales de 6 dígitos (ej: 202405)

        if id_time is None or valor is None or len(str(id_time)) != 6:

            continue

        id_time = str(id_time)

        clave_combinada = (id_time, id_terr)

        # Inicializamos la fila para este mes y esta región si es la primera vez que aparece

        if clave_combinada not in registro_empresas:

            registro_empresas[clave_combinada] = {

                'id_tiempo': id_time,

                'id_territorio': id_terr,

                'empresas_creadas': None,

                'capital_suscrito': None,

                'empresas_disueltas': None

            }

        # 4. Clasificación robusta de métricas (eliminamos problemas de acentos usando búsquedas parciales)

        # --- SOCIEDADES CREADAS / CONSTITUIDAS ---
        print("Antes de constit")
        if "constit" in nombre_lower or "cread" in nombre_lower:

            if "número" in nombre_lower or "numero" in nombre_lower:

                registro_empresas[clave_combinada]['empresas_creadas'] = int(valor)

                print("Estamos dentro del if numero")

            elif "capital" in nombre_lower and "desembolsado" not in nombre_lower:

                registro_empresas[clave_combinada]['capital_suscrito'] = valor

                print("valor")

        # --- SOCIEDADES DISUELTAS ---

        elif "disuelt" in nombre_lower:

            if "número" in nombre_lower or "numero" in nombre_lower:

                registro_empresas[clave_combinada]['empresas_disueltas'] = int(valor)
 
# 5. Volcamos la estructura al diccionario final unificado salud_empresarial

salud_empresarial = {

    'id_tiempo': [],

    'id_territorio': [],

    'empresas_creadas': [],

    'capital_suscrito': [],

    'empresas_disueltas': []

}
 
for reg in registro_empresas.values():

    salud_empresarial['id_tiempo'].append(reg['id_tiempo'])

    salud_empresarial['id_territorio'].append(reg['id_territorio'])

    salud_empresarial['empresas_creadas'].append(reg['empresas_creadas'])

    salud_empresarial['capital_suscrito'].append(reg['capital_suscrito'])

    salud_empresarial['empresas_disueltas'].append(reg['empresas_disueltas'])
 
# 6. Comprobamos el resultado en la terminal

print(f"¡Conseguido! Procesados {len(salud_empresarial['id_tiempo'])} registros combinados mensuales.")
 
# Mostrar una pequeña muestra en pantalla para verificar que no esté vacío

df_resultado = pd.DataFrame(salud_empresarial)

print(df_resultado.head())
 
```

```python
# 1. Cargamos las dimensiones locales que guardasteis en el cuaderno 01

df_territorio = pd.read_csv('territorio.csv')

df_tiempo = pd.read_csv('tiempo.csv')
 
# Reconstruimos los mapas de búsqueda (Nombre -> ID)

map_territorios = {str(row['nombre_territorio']).strip(): int(row['id_territorio']) for _, row in df_territorio.iterrows()}
 
# Creamos un mapa para el tiempo cruzando el año y el FK_Periodo del INE con vuestro id_tiempo mensual

# NOTA: Asegúrate de que tu tiempo.csv tenga las columnas correspondientes (ajusta los nombres si varían)

# Asumimos que mapeamos (anyo, fk_periodo) -> id_tiempo_6_digitos

map_tiempo = {}

for _, row in df_tiempo.iterrows():

    # Ajusta los nombres de las columnas si en tu tiempo.csv se llaman de otra forma (ej. 'anyo', 'id_periodo')

    if 'anyo' in row and 'fk_periodo' in row:

        clave_t = (int(row['anyo']), int(row['fk_periodo']))

        map_tiempo[clave_t] = str(row['id_tiempo'])

    elif 'Anyo' in row and 'FK_Periodo' in row:

        clave_t = (int(row['Anyo']), int(row['FK_Periodo']))

        map_tiempo[clave_t] = str(row['id_tiempo'])
 
registro_empresas = {}
 
# 2. Procesamos el JSON de la variable 'data'

for serie in data:

    nombre_completo = str(serie['Nombre'])

    nombre_lower = nombre_completo.lower()

    # Identificar el territorio

    id_terr = None

    for nombre_terr, id_t in map_territorios.items():

        if nombre_terr.lower() in nombre_lower:

            id_terr = id_t

            break

    if "nacional" in nombre_lower:

        id_terr = map_territorios.get('Nacional')

    if id_terr is None:

        continue

    # 3. Procesamos los datos usando 'Anyo' y 'FK_Periodo'

    for dato in serie['Data']:

        valor = dato.get('Valor')

        anyo = dato.get('Anyo')

        fk_periodo = dato.get('FK_Periodo')

        if valor is None or anyo is None or fk_periodo is None:

            continue

        # Cruzamos con vuestro CSV de tiempo para obtener el ID de tiempo real (ej: '202403')

        clave_tiempo_api = (int(anyo), int(fk_periodo))

        id_time = map_tiempo.get(clave_tiempo_api)

        # Si ese periodo no está en vuestra dimensión de tiempo (o es un dato anual/trimestral que no queréis), lo saltamos

        if id_time is None:

            continue

        clave_combinada = (id_time, id_terr)

        if clave_combinada not in registro_empresas:

            registro_empresas[clave_combinada] = {

                'id_tiempo': id_time,

                'id_territorio': id_terr,

                'empresas_creadas': None,

                'capital_suscrito': None,

                'empresas_disueltas': None

            }

        # 4. Clasificación de las métricas

        if "constit" in nombre_lower or "cread" in nombre_lower:

            if "número" in nombre_lower or "numero" in nombre_lower:

                registro_empresas[clave_combinada]['empresas_creadas'] = int(valor)

            elif "capital" in nombre_lower and "desembolsado" not in nombre_lower:

                registro_empresas[clave_combinada]['capital_suscrito'] = valor

        elif "disuelt" in nombre_lower:

            if "número" in nombre_lower or "numero" in nombre_lower:

                registro_empresas[clave_combinada]['empresas_disueltas'] = int(valor)
 
# 5. Volcamos al diccionario unificado

salud_empresarial = {

    'id_tiempo': [], 'id_territorio': [], 'empresas_creadas': [], 

    'capital_suscrito': [], 'empresas_disueltas': []

}
 
for reg in registro_empresas.values():

    salud_empresarial['id_tiempo'].append(reg['id_tiempo'])

    salud_empresarial['id_territorio'].append(reg['id_territorio'])

    salud_empresarial['empresas_creadas'].append(reg['empresas_creadas'])

    salud_empresarial['capital_suscrito'].append(reg['capital_suscrito'])

    salud_empresarial['empresas_disueltas'].append(reg['empresas_disueltas'])
 
df_resultado = pd.DataFrame(salud_empresarial)

print(f"¡Ahora sí! Procesados {len(df_resultado)} registros coincidentes.")

print(df_resultado.head())
 
```
